```python
import random

class VacuumCleanerAgent:
    def __init__(self, location_a_dirty=True, location_b_dirty=True):
        self.location = random.choice(['A', 'B']) # Start at a random location
        self.environment = {
            'A': 'Dirty' if location_a_dirty else 'Clean',
            'B': 'Dirty' if location_b_dirty else 'Clean'
        }
        print(f"Agent initialized. Start at {self.location}. Environment: {self.environment}")

    def sense(self):
        return self.environment[self.location]

    def act(self, percept):
        if percept == 'Dirty':
            self.environment[self.location] = 'Clean'
            return 'Suck'
        elif self.location == 'A':
            self.location = 'B'
            return 'Move Right'
        else: # self.location == 'B'
            self.location = 'A'
            return 'Move Left'

    def run(self, steps=5):
        print("\n--- Simplified Simulation ---")
        for step in range(1, steps + 1):
            percept = self.sense()
            action = self.act(percept)
            print(f"Step {step}: Agent at {self.location}, Sensed: {percept}, Action: {action}")

# --- Run a single, simplified simulation ---
agent = VacuumCleanerAgent(location_a_dirty=True, location_b_dirty=True)
agent.run(steps=5)
```

```python
import random

class VacuumCleanerGoalAgent:
    def __init__(self, location_a_dirty=True, location_b_dirty=True):
        self.location = random.choice(['A', 'B']) # Start at a random location
        self.environment = {
            'A': 'Dirty' if location_a_dirty else 'Clean',
            'B': 'Dirty' if location_b_dirty else 'Clean'
        }
        self.goal = {'A': 'Clean', 'B': 'Clean'} # The target state
        print(f"Agent initialized. Start at {self.location}. Initial Environment: {self.environment}")
        print(f"Goal: {self.goal}")

    def sense(self):
        return self.environment.copy()

    def is_goal_achieved(self):
        return all(self.environment[loc] == self.goal[loc] for loc in self.goal)

    def act(self, perceived_environment):
        if self.is_goal_achieved():
            return 'NoOp (Goal Achieved)'

        if self.environment[self.location] == 'Dirty':
            self.environment[self.location] = 'Clean'
            return 'Suck'
        else:
            if self.location == 'A':
                self.location = 'B'
                return 'Move Right'
            else: # self.location == 'B'
                self.location = 'A'
                return 'Move Left'

    def run(self, max_steps=10):
        print("\n--- Running Goal-Based Simulation ---")
        for step in range(1, max_steps + 1):
            perceived_environment = self.sense()
            action = self.act(perceived_environment)

            print(f"Step {step}: Agent at {self.location}, Environment: {self.environment}, Action: {action}")

            if action == 'NoOp (Goal Achieved)':
                print(f"Goal achieved at step {step}!")
                break
        else:
            if not self.is_goal_achieved():
                print(f"\nMax steps ({max_steps}) reached. Goal not fully achieved. Final Environment: {self.environment}")
            else:
                print(f"\nGoal achieved at step {max_steps}!")


print("\n--- Scenario 1: Both A and B start Dirty ---")
agent_goal1 = VacuumCleanerGoalAgent(location_a_dirty=True, location_b_dirty=True)
agent_goal1.run(max_steps=10)

print("\n--- Scenario 2: A is Clean, B is Dirty ---")
agent_goal2 = VacuumCleanerGoalAgent(location_a_dirty=False, location_b_dirty=True)
agent_goal2.run(max_steps=10)

print("\n--- Scenario 3: Both A and B start Clean ---")
agent_goal3 = VacuumCleanerGoalAgent(location_a_dirty=False, location_b_dirty=False)
agent_goal3.run(max_steps=10)
```
